In Python , dictionaries are built-in data structures that store collectios of key-value pairs , You should use dictionaries when you need to associate values to unique keys.

STRUCTURE 
dictionary = {
    key1: value1,
    key2: value2
}

First, we find the variable that holds the dictionary. You don't necessarily need to assign the dictionary to a variable, but it's very common to do this to keep it in memory and use it later in the code.

```python
pizza = {
    'name': 'Margherita Pizza',
    'price': 8.9,
    'calories_per_slice': 219,
    'toppings' : ['mozarrella','basil']
}
```

***Accesing dictionary values***
dictionary[key] 

```python
pizza['name']
```

Updating Dictionary Values

```python
pizza['name'] = 'Margherita'
```

Dictionary Methods 
The method get() retrieves the associate value with a keky

```python
pizza.get('toppings')
```

The method key() and values return a visit object with all the keys and valures in the dictionary 

```python
print(pizza.keys())
print(pizza.values())
```

Items() return a list with all key-value pairs in the dictionary 

```python
pizza.items()
```

l método .clear() delete all the key-values pairs:

```python
pizza.clear()
```
