                        **FUNCTIONS**
Functions are reusable pieces of code that run when you call them , the most famous one is the function print()

We can create a new function with the keyword def() followed by the name you want to give your function


**Paramethers** 
The values which we put on the parenthesis and we separate with commas are called paramethers, are the values we pass into functions 

```python
def hello():
    print('Hello world')

hello()

def calculate_sum(a,b):
    print(a+b)

calculate_sum(1,2)
calculate_sum(1,-9)
```
