Sets are one of Python's  built-in data structures. One of the core characteristics of sets is that they don't store duplicate values. If we try to add a duplicate value to a set, only one of them will be stored

Sets are mutable and unordered, which means that their elements are not stored in any specific order, so you cannot use indices or keys to access them. They only can contain values of inmutable data type like numbers , strings and tuples 
And they support mathematical set operations, including union, intersection, difference, and symmetric difference

```python
my_set = {1,2,3,4,5}

print(my_set)
my_set.add(5)
my_set.add(6)
print(my_set)
```
