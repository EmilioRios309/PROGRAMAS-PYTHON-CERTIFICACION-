***Tuples are very similar to lists , lists are a mutable data typle , tuples are inmutable. This means that the elements in a tuple cannot be changed once it's created ***


One of the most famous useful uses of tuples is unpack items from a tuple just like you did with lists

```python
developer = ('Alicia', 34, 'Rust Developer')
name, age, job = developer
print(name,age,job,sep='\n')
```
