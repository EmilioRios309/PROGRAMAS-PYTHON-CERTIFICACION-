Attributes are variables that belongs to an object , so they hold data. There are two kinds of attributes : instance attributes and class attributes 
Instance attributes are unique to each object created from a class, and you usually set them with the __init__ method. Class attributes, on the other hand, belong to the class itself and are shared by all instances of that class

```python
class Dog:
    specie = "French Buldog" #Class atribute

    def __init__(self,name):
        self.name = name

print(Dog.specie)
dog1 = Dog("Jorge")
print(dog1.specie)
print(dog1.name)
```

```python
class Car:
    def __init__(self,modelo,color):
        self.modelo = modelo
        self.color = color

car1 = Car("Toyota","Verde")
car2 = Car("Nissan","Gris")

print(car1.modelo,car1.color)
print(car2.modelo,car2.color)
```

Methods are functions defined inside a class. With them, any object defined from a class can perform actions that operate on or modify its own data. You also access a method with dot notation.

```python
class Dog:
    def __init__(self, name):
        self.name = name

    def bark(self):
        print(f"{self.name} says woof woof")

        
```
