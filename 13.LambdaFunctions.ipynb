We am used to define function using the keyword def 

```python
def square(num):
    return num **2

print(square(5))
```

But has we learn when we are using some funcitions as filter and map() we can use a anonymous line , and here's where the lamda functions works 

```python
lambda num: num **2
```

```python
#Some examples substituing 
numbers = [1,2,3,4,5]
even_numbers = list(filter(lambda num: num%2 == 0,numbers))
print(even_numbers)
```

Is not  a good practice assing a function lambda in a variable 

```python
square = lambda num: num **2
squares = list(map(square,numbers))
print(squares)
```
