***Common Methods***

```python
numbers = [1, 2, 3, 4 ,5]
numbers.append(6)
print(numbers)

even_numbers = [7,9,10]
numbers.append(even_numbers)
print(numbers)
```

But if you want to add all of the individual numbers from the even_numbers list at the end of the numbers list, then you can use the extend() method.The extend() method is similar to the append() method, but with extend() you can add multiple elements from one list to another.

```python
numbers = [1, 2, 3, 4, 5]
even_numbers = [6,8,10]
numbers.extend(even_numbers)
print(numbers)
```

To insert an element at a specific index in a list, you can use the insert() method. This method accepts two arguments: the index where you wish to insert the new item and the item you want to insert.

```python
numbers.insert(2,2.5)
print(numbers)
```

If you want to remove an element from a list, you can use the remove() method. The remove() method takes the value of the element to remove as an argument:

```python

print(numbers)
```

To remove an element at a specific index in the list, you can use the pop() method like this:



```python
numbers.pop(1)
print(numbers)
```
