Python provides the **try**, **except**, **finally** blocks to gracefully handle errors


```python
try:
    x = 10/0
except ZeroDivisionError:
    print("You can't divide by Zero")
    
```

```python
try:
    x = 10/2
except ZeroDivisionError:
    print("You can't divide by zero")
else:
    print('Division succesful',x)
finally:
    print('This always prints')

```
