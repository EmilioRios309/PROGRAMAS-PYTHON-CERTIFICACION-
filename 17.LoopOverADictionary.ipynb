Loop over dictionary values

```python
products = {
    'Laptop': 10000,
    'Celular': 12000,
    'Audifonos' : 13000,
    'Rayador': 8,
}

for price in products.values(): #Loop over values
    print(price)
```

```python
for product in products.keys():
    print(product) #Loop Over keys 
```

```python
#Loop over key-value pairs
for product,price in products.items():
    print(f'{product}: {price}')
```

Using enumerate () to Track a Counter   
We can loop over a key-value pair while keeping track of a counter , you can call enumerate() function . This counter essentially acts as a sort of "index" or "count"


```python
for double in enumerate(products): #We can assign two variable to unpackage all the tuples
    print(double)

print()

for double in products.items():
    print(double)
```

```python
for index, product in enumerate(products):
    print(f'{index}: {product}')

print()

#Loop over prices enumerating 
for index,price in enumerate(products.values(),1):
    print(f"{index}: {price}")
```
