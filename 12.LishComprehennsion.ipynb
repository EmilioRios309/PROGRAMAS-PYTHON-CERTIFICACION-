```python
even_numbers = list()

for num in range(21):
    if num % 2 == 0 :
        even_numbers.append(num)

print(even_numbers)
```

***List Comprehension***
While this code works, there is a more concise way to write this that uses list comprehension instead. List comprehension allows you to create a new list in a single line by combining a loop and condition directly within square brackets 
This makes the code shorter and often easier to read.


```python
even_numbers = [num for num in range(21) if num % 2 == 0]
even_numbers
```

```python
numbers = [1,2,3,4,5,6]
result = [(num,'Even') if num  %2 == 0 else (num, 'Odd') for num in numbers]
print(result)
```

Another way to make a list from an existent iterable is the function filter()

The function filter() is used to select elements of an iterable that meet a specific conditon . Filter accepts a function and an iterable as arguments 

```python
words = ['Python','JavaScript','Rust','C++','Java','TypeScript']

def is_long_word(word):
    return len(word) > 4

longer_words = list(filter(is_long_word,words))
print(longer_words)
```

THere are some functions that are useful when we work with lists. Another function that we must to understand is the function map() that takes an iterable and aplies a function in each element of the iterable    

```python
celsius = [0,10,20,30,40,50]

def to_farenhaint(temp):
    return (temp * 9/5) + 32

faremhaint = list(map(to_farenhaint,celsius))
print(faremhaint)
```
