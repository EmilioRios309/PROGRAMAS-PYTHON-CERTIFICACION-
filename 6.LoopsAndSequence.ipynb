**LISTS AND HOW THEY WORK**

The list data type is an ordered sequence of elements that can be comprised of strings, numbers, or even other lists. The first element of the list is at index zero


```python
listas = []
cities = ['Los Angeles', 'Mexico City', 'Madrid ']
cities[0]
```

Negative Index is used to access at the final of the list or convert a string in a list 


```python
developer = 'Jessica'
list(developer)
```

```python
numbers = [1, 2, 3, 4, 5]
len(numbers)
```

If you wanted to update a value at a particular index, you can do something like this:

```python
programming_language = ['Python', 'Java', 'C++',  'Rust']
programming_language[0] = 'JavaScript'
programming_language
```

```python
del programming_language[1]
programming_language
```

Sometimes it is helpful to check if an element is inside the list. To do that, you can use the in keyword like this:

```python
print('Rust' in programming_language)
print('Java' in programming_language)
```
